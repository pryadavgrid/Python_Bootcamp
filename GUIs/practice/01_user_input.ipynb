```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Create input fields
name = widgets.Text(
    description='Name:'
)

email = widgets.Text(
    description='Email:'
)

phone = widgets.Text(
    description='Phone:'
)

# Submit button
submit_btn = widgets.Button(
    description='Submit',
    button_style='success'
)

# Output area
output = widgets.Output()

# Function when button clicked
def on_submit_clicked(b):
    with output:
        clear_output()
        print("Form Submitted Successfully ✅\n")
        print("Name  :", name.value)
        print("Email :", email.value)
        print("Phone :", phone.value)

submit_btn.on_click(on_submit_clicked)

# Display all
display(name, email, phone, submit_btn, output)
```
